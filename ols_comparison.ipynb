```python
import os
import torch
import torch.nn.functional as F

def load_tensors_from_folder(folder_path):
    tensors = []
    for filename in os.listdir(folder_path):
        if filename.endswith('.pt'):
            tensor = torch.load(os.path.join(folder_path, filename))
            tensors.append(tensor)
    return tensors

# Load tensors from both folders
whatsapp_tensors = load_tensors_from_folder('heatmaps/good_wind_7/Pne_Nor_Cov_baseline-lr:0.001-wd:0.0_good-wind-7/whatsapp')
original_tensors = load_tensors_from_folder('heatmaps/good_wind_7/Pne_Nor_Cov_baseline-lr:0.001-wd:0.0_good-wind-7/original')
```

```python
def compute_cosine_similarity(tensor1, tensor2):
    tensor1 = tensor1.flatten()
    tensor2 = tensor2.flatten()
    cosine_similarity = F.cosine_similarity(tensor1.unsqueeze(0), tensor2.unsqueeze(0))
    return cosine_similarity.item()

# Compute cosine similarities
```

```python
def compute_kl(tensor1,tensor2):
    tensor1 = tensor1.flatten()
    tensor2 = tensor2.flatten()
    kl = F.kl_div(tensor1.log(), tensor2, reduction='batchmean')
    return kl.item()
```

```python
# Flatten the tensors
tensor1 =  data_og
tensor2 = data_per
tensor1_flat = tensor1.flatten()
tensor2_flat = tensor2.flatten()

# Normalize tensors to be valid probability distributions (they should sum to 1)
tensor1_prob = tensor1_flat / torch.sum(tensor1_flat)
tensor2_prob = tensor2_flat / torch.sum(tensor2_flat)

# Add a small epsilon to avoid log(0)
epsilon = 1e-10
tensor1_prob = torch.clamp(tensor1_prob, epsilon, 1)
tensor2_prob = torch.clamp(tensor2_prob, epsilon, 1)

# Compute the KL divergence
kl_div = torch.nn.functional.kl_div(tensor1_prob.log(), tensor2_prob, reduction='batchmean')
print(kl_div.item())
```

```python
og = torch.stack(original_tensors)
wa = torch.stack(whatsapp_tensors)
og = og.unsqueeze(1)
wa =wa.unsqueeze(1)
og.shape
```

```python
from ignite.metrics import SSIM
from ignite.engine import Engine
def process_function(engine, batch):
    y_pred, y_true = batch
    return y_pred, y_true
evaluator = Engine(process_function)
ssim = SSIM(data_range=1.0)
ssim.attach(evaluator, "ssim")
state = evaluator.run([[og, wa]])
print("SSIM:", state.metrics['ssim'])
```

```python
def compute_avg_similarity(whatsapp_tensors,original_tensors):
    cosine_similarities = []
    for whatsapp_tensor, original_tensor in zip(whatsapp_tensors, original_tensors):
        similarity = compute_cosine_similarity(whatsapp_tensor, original_tensor)
        cosine_similarities.append(similarity)
    average_cosine_similarity = sum(cosine_similarities) / (len(cosine_similarities) + 1e-6)
    print(f'Average Cosine Similarity: {average_cosine_similarity}')
```

```python
def compute_avg_kl(whatsapp_tensors,original_tensors):
    kl = []
    for whatsapp_tensor, original_tensor in zip(whatsapp_tensors, original_tensors):
        kl_div = compute_kl(whatsapp_tensor, original_tensor)
        kl.append(kl_div)
    average_kl = sum(kl) / (len(kl) + 1e-6)
    print(f'Average KL Divergence: {average_kl}')
```

```python
compute_avg_similarity(whatsapp_tensors,original_tensors)
compute_avg_kl(whatsapp_tensors,original_tensors)
```
