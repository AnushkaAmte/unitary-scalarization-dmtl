```python
rho_df = pd.read_csv('chexphoto/test_results/chexphoto0_7_13_14_r:1_baseline-lr:0.001-wd:0.0_still-galaxy-142_last.csv')
grad_df = pd.read_csv('heatmaps/csvfiles/grad_cam/ols_0_7_13_14_r:1_baseline-lr:0.001-wd:0.0_still-galaxy-142.csv')
norm_df = pd.read_csv('heatmaps/csvfiles/grad_norm_sq/ols_0_7_13_14_r:1_baseline-lr:0.001-wd:0.0_still-galaxy-142.csv')
```

```python

prefix = '1_COVID-19/'
rho_df['image_name'] = rho_df['image_name'].str.replace(prefix, '', regex=False)

rho_df
```

```python
grad_df = grad_df.drop(columns=['Unnamed: 0'])
norm_df = norm_df.drop(columns=['Unnamed: 0'])
```

```python
#rename columns
grad_df = grad_df.rename(columns={"Image_Name": "image_name"})
norm_df = norm_df.rename(columns={"Image_Name": "image_name"})
```

```python
grad_combine = pd.merge(grad_df, norm_df, on='image_name',how='outer',suffixes=('_grad', '_norm'))
final_df = pd.merge(rho_df, grad_combine, on='image_name',how='outer')
```
